```python
import pandas as pd
import os
from fredapi import Fred
fred = Fred(api_key = os.getenv("FRED_API_KEY"))
```

```python
interest_rate = fred.get_series('FEDFUNDS')
mortgage_rate = fred.get_series('MORTGAGE30US')
#consumption = fred.get_series('PCEC96')
#investments = fred.get_series('PNFIC1')
unemployment_rate = fred.get_series('UNRATE')
payrolls = fred.get_series('PAYEMS')
labor_turnover = fred.get_series('JTSJOL')
cpi = fred.get_series('CPIAUCSL')
cepi = fred.get_series('DPCCRV1Q225SBEA')
ppi = fred.get_series('PPIACO')
retail_sales = fred.get_series('MRTSSM44000USS')
personal_income = fred.get_series('PI')
personal_saving = fred.get_series('PMSAVE')
personal_spending = fred.get_series('PCE')
housing_starts = fred.get_series('MSPNHSUS')
gdp = fred.get_series('GDP')
industrial_prod = fred.get_series('INDPRO')
#nfci = fred.get_series('NFCICREDIT')
money_stock = fred.get_series('M1REAL')
#bank_lending = fred.get_series('TOTLL')
trade_balance = fred.get_series('BOPGSTB')
exports = fred.get_series('NETEXP')
imports = fred.get_series('IMPGS')
consumer_sentiment = fred.get_series('UMCSENT')
vehicle_sales = fred.get_series('TOTALSA')
```

```python
df = pd.concat([interest_rate, unemployment_rate, payrolls, labor_turnover, cpi, cepi, ppi, retail_sales, personal_income, personal_saving, personal_spending, housing_starts, gdp, industrial_prod, money_stock, trade_balance, exports, imports, consumer_sentiment, vehicle_sales], axis =1)
df.columns = ['interest_rate', 'unemployment_rate', 'payrolls', 'labor_turnover','cpi', 'cepi', 'ppi', 'retail_sales', 'personal_income', 'personal_saving', 'personal_spending', 'housing_starts', 'gdp','industrial_prod','money_stock', 'trade_balance', 'exports', 'imports', 'consumer_sentiment', 'vehicle_sales']
df=df.dropna()
df.describe()
```
